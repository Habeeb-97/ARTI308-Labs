# Lab 7 - Logistic Regression

In this lab, we apply Logistic Regression on a new dataset (Students Performance).

The goal is to classify students into:
- Pass (1)
- Fail (0)

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
```

## Loading the Dataset

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

## Exploring the Data

```python
df.info()
df.describe()
```

## Data Cleaning
We check for missing values and duplicates.

```python
print(df.isnull().sum())
print(df.duplicated().sum())

df = df.drop_duplicates()
df = df.dropna()
```

## Feature Engineering
We convert categorical variables into numeric using one-hot encoding.

```python
df = pd.get_dummies(df, drop_first=True)
df.head()
```

## Creating Target Variable

We create a new column called 'pass':
- 1 if math score >= 50
- 0 otherwise

```python
df['pass'] = df['math score'].apply(lambda x: 1 if x >= 50 else 0)
df.head()
```

## Preparing Data for Modeling

```python
X = df.drop(['math score', 'pass'], axis=1)
y = df['pass']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

## Training the Model

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

## Making Predictions

```python
predictions = model.predict(X_test)
```

## Model Evaluation
We evaluate the model using:
- Confusion Matrix
- Classification Report

```python
print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions))
```

## Visualization

```python
sns.heatmap(confusion_matrix(y_test, predictions), annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.show()
```

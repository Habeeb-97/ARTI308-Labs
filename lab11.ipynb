# Lab 11 - K-Means Clustering

In this lab, we apply K-Means clustering on the Students Performance dataset.

The goal is to group students into clusters based on their scores and performance-related features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

## Loading the Dataset

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

## Exploring the Data

```python
df.info()
df.describe()
```

## Data Cleaning

```python
print(df.isnull().sum())
print(df.duplicated().sum())

df = df.drop_duplicates()
df = df.dropna()
```

## Feature Engineering

We convert categorical columns into numeric values using one-hot encoding.

df_encoded = pd.get_dummies(df, drop_first=True)
df_encoded.head()

## Feature Scaling

K-Means depends on distances, so scaling is important.

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df_encoded)
```

## Finding the Best Number of Clusters

We use the Elbow Method to choose a suitable value for K.

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

## Training the K-Means Model

Based on the elbow method, we choose 3 clusters.

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(scaled_data)
```

## Adding Cluster Labels to the Dataset

```python
df['Cluster'] = clusters
df.head()
```

## Cluster Analysis

```python
df.groupby('Cluster')[['math score', 'reading score', 'writing score']].mean()
```

## Visualization

```python
sns.scatterplot(
    x='math score',
    y='reading score',
    hue='Cluster',
    data=df,
    palette='viridis'
)

plt.title('Student Clusters')
plt.show()
```

## Conclusion

K-Means clustering was used to group students based on their performance.  
The clusters show different student groups according to their scores and features.
